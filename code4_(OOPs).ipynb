Classes and objects

Object oriented programming is a programming paradigm that uses "objects" to design application and computer programs. OOP allows us for modeling real world scenarios using classes and objects.

```python
class Dog:
    # Constructor
    def __init__(self, name, age):
        self.name = name
        self.age = age

dog1 = Dog("Buddy", 3)

print(dog1.name)
print(dog1.age)
```

```python
# define a class with instance methods
class dog:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def bark(self):
        print(f"{self.name} says woof")

dog1=dog("buddy",3)
dog1.bark()
dog2=dog("lucy",4)
dog2.bark()
```

```python
class BankAccount:

    def __init__(self, name, balance=0):
        self.name = name
        self.balance = balance

    def deposit(self):
        self.balance += int(input("Deposit Amount: ₹"))

    def withdraw(self):
        amount = int(input("Withdraw Amount: ₹"))
        if amount <= self.balance:
            self.balance -= amount
            print("Withdraw Successful")
        else:
            print("Insufficient Balance")

    def check_balance(self):
        print(f"Current Balance: ₹{self.balance}")


account = BankAccount("Navneet")

account.deposit()
account.withdraw()

if input("Check Balance? (yes/no): ").lower() == "yes":
    account.check_balance()
else:
    print("Thank You!")
```

Inheritance in python

Inheritance is a fundamental concept in object oriented programming that allows a class to inherit attributes and methods from another class.

```python
# inheritance ( single )
# parent
class Car:
    def __init__(self, windows, doors, enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype

    def drive(self):
        print(f"The person will drive the {self.enginetype} car")
        
```

```python
# child
class Tesla(Car):
    def __init__(self, windows, doors, enginetype, is_selfdrive):
        super().__init__(windows, doors, enginetype)
        self.is_selfdrive = is_selfdrive

    def selfdrive(self):
        print(f"Tesla supports self drive: {self.is_selfdrive}")
```

```python
tesla1 = Tesla(4, 5, "electric", True)

tesla1.drive()
tesla1.selfdrive()
```

# multiple inheritance
# when a class inherits from more than one base class

```python
# base class 1( parent 1)
class Animal:
    def __init__(self,name):
        self.name=name

    def speak(self):
        print("subclass must implement this mrthod")

# base class 2 (parent 2)
class pet:
     def __init__(self,owner):
            self.owner=owner

# Derived class (child)
class dog(Animal,pet):
    def __init__(self, name,owner):
         Animal.__init__(self,name)
         pet.__init__(self,owner)

    def speak(self):
         return f"{self.name} say woof"

# obj
dog1 = dog('buddy','navneet')
print(dog1.speak())
print(f"owner: {dog1.owner}")
```

polymorphism 

polymorphism is a core concept in object oriented programming that allows object of different class to be treated as objects of a common super class. It Provides you a way to perform a single action in different forms . Polymer fission is typically achieved through method overriding and interface .

method overriding
Method overriding allows a children class to provide a specific implementation of a method that is already defined in its parent class .

```python
## Base Class

class Animal:
    def speak(self):
        return "Sound of the animal"

## Derived Class 1

class Dog(Animal):
    def speak(self):
        return "Woof!"

## Derived class

class Cat(Animal):
    def speak(self):
        return "Woof!"

## Derived class

class Cat(Animal):
    def speak(self):
        return "Meow!"

# obj
dog=Dog()
cat=Cat()
print(dog.speak())
print(cat.speak())
```
