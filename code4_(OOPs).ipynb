Classes and objects

Object oriented programming is a programming paradigm that uses "objects" to design application and computer programs. OOP allows us for modeling real world scenarios using classes and objects.

```python
class Dog:
    # Constructor
    def __init__(self, name, age):
        self.name = name
        self.age = age

dog1 = Dog("Buddy", 3)

print(dog1.name)
print(dog1.age)
```

```python
# define a class with instance methods
class dog:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def bark(self):
        print(f"{self.name} says woof")

dog1=dog("buddy",3)
dog1.bark()
dog2=dog("lucy",4)
dog2.bark()
```

```python
class BankAccount:

    def __init__(self, name, balance=0):
        self.name = name
        self.balance = balance

    def deposit(self):
        self.balance += int(input("Deposit Amount: ₹"))

    def withdraw(self):
        amount = int(input("Withdraw Amount: ₹"))
        if amount <= self.balance:
            self.balance -= amount
            print("Withdraw Successful")
        else:
            print("Insufficient Balance")

    def check_balance(self):
        print(f"Current Balance: ₹{self.balance}")


account = BankAccount("Navneet")

account.deposit()
account.withdraw()

if input("Check Balance? (yes/no): ").lower() == "yes":
    account.check_balance()
else:
    print("Thank You!")
```

Inheritance in python

Inheritance is a fundamental concept in object oriented programming that allows a class to inherit attributes and methods from another class.

```python
# inheritance ( single )
# parent
class Car:
    def __init__(self, windows, doors, enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype

    def drive(self):
        print(f"The person will drive the {self.enginetype} car")
        
```

```python
# child
class Tesla(Car):
    def __init__(self, windows, doors, enginetype, is_selfdrive):
        super().__init__(windows, doors, enginetype)
        self.is_selfdrive = is_selfdrive

    def selfdrive(self):
        print(f"Tesla supports self drive: {self.is_selfdrive}")
```

```python
tesla1 = Tesla(4, 5, "electric", True)

tesla1.drive()
tesla1.selfdrive()
```

polymorphism 

polymorphism is a core concept in object oriented programming that allows object of different class to be treated as objects of a common super class. It Provides you a way to perform a single action in different forms . Polymer fission is typically achieved through method overriding and interface .

method overriding
Method overriding allows a children class to provide a specific implementation of a method that is already defined in its parent class .

```python
## Base Class

class Animal:
    def speak(self):
        return "Sound of the animal"

## Derived Class 1

class Dog(Animal):
    def speak(self):
        return "Woof!"

## Derived class

class Cat(Animal):
    def speak(self):
        return "Woof!"

## Derived class

class Cat(Animal):
    def speak(self):
        return "Meow!"

# function that demonstate polymorphism
def animal_speak(animal):
    print(animal.speak())


# obj
dog=Dog()
cat=Cat()
print(dog.speak())
print(cat.speak())
print(dog.speak())
```

```python
# Hierarchical Inheritance
# polymorshism with function and methods
# base class
class shape():
    def area(self):
        return 'the area of the figure'

# derived class 1
class Rectangle(shape):
    def __init__(self,width,height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

# derived class 2
class Circle(shape):
    def __init__(self,radius):
        self.radius = radius

    def area(self):
        return self.radius * self.radius * 3.14

# function that demonstrate polymorphism
def print_area(shape):
    print(f"the area id {shape.area()}")

# obj
rectangle = Rectangle(4,5)
circle = Circle(3)

print_area(rectangle)
print_area(circle)
        
```

# polymorphism with abstract base class 
Abstract base classes (ABCs) are used to define common method for a group of related objects. They can enforce that derived class implement a particular methods, promoting consistency across different implementation .

```python
from abc import ABC, abstractmethod
# define as abstract class
class Vehicle(ABC):
    @abstractmethod 
    def start_engine(self):
        pass

# derived class 1
class Car(Vehicle):
    def start_engine(self):
        return "Car engine started"

# drived class 2
class Motorcycle(Vehicle):
    def start_engine(self):
        return "Motorcycle engine started"


# Function
def start_vehicle(vehicle):
    print(vehicle.start_engine())

# obj
car = Car()
motocycle = Motorcycle()

start_vehicle(car)
start_vehicle(motocycle)
```

# Encapsulation
Encapsulation is the concept of wrapping data (variables) and methods (functions) together a single  unit. It resticts direct access to some of the objects components, which is a mean of preventing accidental interface and misuse of the data.

```python
# Encapsulation with Getter and setter methods
# public,protected,private variable or access modifyer

class Person:
    def __init__(self,name,age):
        self.name =name             # public variable
        self.age = age              # public variable

def get_name(person):
    return person.name

person = Person("navneet",34)
print(person.name)
print(person.age)
get_name(person)
```

```python


class Person:
    def __init__(self,name,age,gender):
        self.__name =name             # private variable
        self.__age = age              # private variable
        self.gender = gender          # public variable

person = Person("navneet",34,"male")

```

```python
class Person:
    def __init__(self,name,age,gender):
        self._name =name             # protected variable
        self._age = age              # protected variable
        self.gender = gender          # public variable

class Employee(Person):
    def __init__(self, name, age, gender):
        super().__init__(name, age, gender)

employee = Employee("navneet",34,"male")
print(employee._name)

```

Abstraction

Abstraction is a concept of hiding the complex implementation detail and showing only the necessary feature of an object this helps in reducing programming complexity and effort .

```python
from abc import ABC, abstractmethod
# abstract base class 
class Vehicle(ABC):
    def drive(self):
        print("the vehicle is used for driving")

    @abstractmethod
    def start_engine(self):
        pass

class car(Vehicle):
    def start_engine(self):
        print("car engine started")

def operate_vehicle(vehicle):
    vehicle.start_engine()
    vehicle.drive()

#obj
car = Car()
operate_vehicle(car)
```

overriding operation

```python
class Vector:
    def __init__(self,x,y):
        self.x =x
        self.y =y

    def __add__(self, other):
        return Vector(self.x+other.x,self.y + other.y)

    def __sub__(self, other):
            return Vector(self.x - other.x,self.y - other.y)

    def __repr__(self):
        return f" Vector ({self.x},{self.y})"
# obj
V1 = Vector(2,5)
V2 = Vector(7,5)
print(V1 + V2)
print(V1 - V2)
```
