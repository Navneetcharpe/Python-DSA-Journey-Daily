Bubble sort

```python
# O(n2)


def bubblesort(a):
    n = len(a)

    for i in range(n):
        for j in range(0, n-1-i):
            if(a[j]  > a[j+1]):
              a[j],a[j+1] = a[j+1],a[j]  
    
a= [64.27,47,24,23,54,53]
bubblesort(a)
print(a)
```

selection sort

```python
# O(n2)

def selectionsort(a):
    n = len(a)

    for i in range(n):
        min = i
        for j in range(i, n):
            if(a[min] > a[j]):
                min = j

        a[i],a[min] = a[min],a[i]

a= [64.27,47,24,23,54,53]
selectionsort(a)
print(a)
```

insertion sort

```python
def insertionsort(a):
    n =len(a)

    for i in range(1,n):
        key = a[i]
        j= i -1
        while(j>=0 and key <a[j]):
            a[j+1] = a[j]
            j = j-1

        a[j+1] = key

a= [64.27,47,24,23,54,53]
insertionsort(a)
print(a)
```
