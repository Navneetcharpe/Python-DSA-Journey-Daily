Bubble sort

```python
# O(n2)


def bubblesort(a):
    n = len(a)

    for i in range(n):
        for j in range(0, n-1-i):
            if(a[j]  > a[j+1]):
              a[j],a[j+1] = a[j+1],a[j]  
    
a= [64.27,47,24,23,54,53]
bubblesort(a)
print(a)
```

selection sort

```python
# O(n2)

def selectionsort(a):
    n = len(a)

    for i in range(n):
        min = i
        for j in range(i, n):
            if(a[min] > a[j]):
                min = j

        a[i],a[min] = a[min],a[i]

a= [64.27,47,24,23,54,53]
selectionsort(a)
print(a)
```

insertion sort

```python
def insertionsort(a):
    n =len(a)

    for i in range(1,n):
        key = a[i]
        j= i -1
        while(j>=0 and key <a[j]):
            a[j+1] = a[j]
            j = j-1

        a[j+1] = key

a= [64.27,47,24,23,54,53]
insertionsort(a)
print(a)
```

Merge sort

```python
# DIvide and conqeuere

# O(n log n)

def Divide(arr, l,r):
    if (l<r):
        m = (l+r) //2
        Divide(arr,l,m)
        Divide(arr,m+1,r)
        merge(arr,l,m,r)

def merge(arr,l,m,r):
    S1 = m-l +1
    S2 = r - m

    L = [0] * S1
    R = [0] * S2

    for i in range(S1):
        L[i]=arr[l+i]

    for j in range(S2):
        R[j]=arr[m+1 + j]

    i = j = 0
    k = l

    while(i< S1 and j <S2):
        if(L[i] < R[j]):
            arr[k] = L[i]
            i = i+1
            k=k+1
        else:
            arr[k] = R[j]
            j=j+1
            k=k+1

    while(i< S1):
        arr[k] = L[i]
        i = i+1
        k=k+1

    while(j< S2):
        arr[k] = R[j]
        j = j+1
        k=k+1

arr = [21,34,33,66,34,2,24]
Divide(arr,0,len(arr)-1)
print(arr)
```

Quick sort

```python
# Divide and conqeuere
# O(n log n)

# O(n2) werst case

def Quicksort(arr,l,r):
    if(l<r):
        p = partition(arr,l,r)

        Quicksort(arr,l,p-1)
        Quicksort(arr,p+1,r)

def partition(arr,l,r):
    pivot = arr[l]
    i =l+1
    j = r

    while True:
        while i <= r and arr[i] < pivot:
            i = i + 1

        while j >= l and arr[j] > pivot:
            j = j - 1

        if(i < j):
            arr[i],arr[j] = arr[j],arr[i] 
        else:
            break

    arr[l],arr[j] = arr[j],arr[l]
    return j

arr = [25,23,654,76,4,63,43,5]
Quicksort(arr,0,len(arr)-1)
print(arr)
```

```python
def find_min_max(arr, start, end):
    
    if start == end:
        return arr[start], arr[start]

    if start + 1 == end:
        if arr[start] < arr[end]:
            return arr[start], arr[end]
        else:
            return arr[end], arr[start]

    mid = (start + end) // 2

    min1, max1 = find_min_max(arr, start, mid)
    min2, max2 = find_min_max(arr, mid + 1, end)

    return min(min1, min2), max(max1, max2)


arr = [23, 43, 63, 22, 43, 64]

minimum, maximum = find_min_max(arr, 0, len(arr) - 1)

print("min:", minimum)
print("max:", maximum)
```
