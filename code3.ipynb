```python
# linear search
def  linear_search (arr,target):
    size = len(arr)
    for index in range(0,size):
        if (arr[index]==target):
            return index
    return -1

lists=[14,4,65,234,2,4,23,56,34,5]
target = 23

result= linear_search(lists,target)
print(result)
```

```python
# Binary search

def binary_search(arr,target):
    size= len(arr)

    start=0
    end=size -1

    while (start <= end):
        mid = (start + end)//2

        if(arr[mid]== target):
            return mid
        elif (arr[mid]> target):
            end =mid -1
        elif (arr[mid]< target):
            start =mid +1
    return -1

sorted_list = [10,20,44,23,54,38,34,23,22,53,35,83]
target=22 

result= binary_search(sorted_list,target)
print(result)
```

```python
# sorting algorithm
# bubble sort
def bubble_sort(arr):
    size = len(arr)

    for passes in range(0,size):
        for j in range(0,size-1-passes):
            if (arr[j]>arr[j+1]):
                arr[j],arr[j+1] = arr[j+1],arr[j]
    return arr


unsorted_array = [12,34,54,98,23,56,35,75,23]
sorted_list = bubble_sort(unsorted_array)
print("sorted element",sorted_list)
```

```python
# selection sort (one position sort array)
def selection_sort(arr):
     size = len(arr)

     min_index=0

     for j in range(1,size):
          if (arr[j]<arr[min_index]):
                min_index= j

     arr[0],arr[min_index]= arr[min_index],arr[0]

     return arr

unsorted_array = [65,34,54,14,35,75,23]
sorted_list = selection_sort(unsorted_array)
print("sorted element",sorted_list)

#  or (sorted list)

def selection_sort(arr):
     size = len(arr)
     for i in range(size-1):
          min_index=i

     for j in range(i+1,size):
          if (arr[j]<arr[min_index]):
                min_index= j

     arr[i],arr[min_index]= arr[min_index],arr[i]

     return arr

unsorted_array = [86,34,54,14,35,75,23]
sorted_list = selection_sort(unsorted_array)
print("sorted element",sorted_list)


```

```python
# insertion sort 
def insertion_sort(arr):
    n = len(arr)


    for current in range(1,n):
        currentcard =arr[current]
        correntpositon = current-1
        while correntpositon >=0:
            if (arr[correntpositon] < currentcard):
                break
            else:
                arr[correntpositon +1] = arr[correntpositon]
                correntpositon -= 1
            arr[correntpositon +1]= currentcard
    return arr

unsorted_array = [34,56,23,56,78,23,45,67]
sorted_list = insertion_sort(unsorted_array)
print("sorted element",sorted_list)
```

File Opration :- read and write files 

File handling is a circular path of any programming language. Python provider in build function and methods to read from a right to file, we both text and binary.

```python
# read a whole file
with open('package/source.txt','r')as file:
    content=file.read()
    print(content)
```

```python
# read a file line by line
with open('package/source.txt','r')as file:
    for line in file:
        print(line.strip()) # strip() removes the newlines character
```

```python
#writing a file(overwriting)
with open('package/source.txt','w')as file:
    file.write('hello world\n')
    
```

```python
#writing a file(without overwriting)
with open('package/destination.txt','a')as file:
    file.write('\nappend operation taking place')
```

```python
# writing a list of line to a  flie
lines = ['\nfirst line \n','second line \n','third line \n']
with open('package/destination.txt','a')as file:
    file.writelines(lines)
```

```python
# Binary files
# write to a binary file
data = b'\x00\x01\x02\x03\x04'
with open('package/ex.bin','wb')as file:
    file.write(data)

with open('package/ex.bin','rb')as file:
    content = file.read()
    print(content)
```

```python
# read the content from a source text file and write it to a and destination text file
with open('package/destination.txt','r')as destination_file:
   content = destination_file.read()

with open('package/source.txt','w')as source_file:
    source_file.write(content)



```
