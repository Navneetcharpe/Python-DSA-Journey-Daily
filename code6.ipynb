```python
class Node:
    def __init__(self, info, next=None):
        self.data = info
        self.next = next


class SinglyLinkedlist:
    def __init__(self, head=None):
        self.head = head

    def insertAtEnd(self, value):
        temp = Node(value)

        if self.head != None:
            t1 = self.head

            while t1.next != self.head:
                t1 = t1.next

            t1.next = temp
            temp.next = self.head

        else:
            self.head = temp
            temp.next = self.head

    def insertAtbeg(self, value):
        temp = Node(value)

        if self.head == None:
            self.head = temp
            temp.next = self.head
        else:
            t1 = self.head

            while t1.next != self.head:
                t1 = t1.next

            temp.next = self.head
            self.head = temp
            t1.next = self.head

    def insertAtMid(self, value, x):
        temp = Node(value)

        if self.head == None:
            return

        t1 = self.head

        while True:
            if t1.data == x:
                temp.next = t1.next
                t1.next = temp
                return

            t1 = t1.next

            if t1 == self.head:
                break

    def deleteLL(self, value):
        if self.head == None:
            return

        # Only one node
        if self.head.data == value and self.head.next == self.head:
            self.head = None
            return

        # Delete head
        if self.head.data == value:
            t1 = self.head

            while t1.next != self.head:
                t1 = t1.next

            self.head = self.head.next
            t1.next = self.head
            return

        # Delete other node
        prev = self.head
        t1 = self.head.next

        while t1 != self.head:
            if t1.data == value:
                prev.next = t1.next
                return

            prev = t1
            t1 = t1.next

    def printLL(self):
        if self.head == None:
            return

        t1 = self.head

        while True:
            print(t1.data)
            t1 = t1.next

            if t1 == self.head:
                break


obj = SinglyLinkedlist()

obj.insertAtEnd(10)
obj.insertAtEnd(20)
obj.insertAtEnd(50)
obj.insertAtEnd(170)
obj.insertAtEnd(140)

obj.insertAtbeg(60)

obj.insertAtMid(120, 50)

obj.deleteLL(170)

obj.printLL()
```

```python

class Node:
    def __init__(self, value=None):
        self.data = value
        self.next = None
        self.prev = None


class DoublyLL:
    def __init__(self):
        self.head = None

    def insertAtEnd(self, value):
        temp = Node(value)

        if self.head == None:
            self.head = temp
            temp.next = self.head
            temp.prev = self.head
            return

        t = self.head

        while t.next != self.head:
            t = t.next

        t.next = temp
        temp.prev = t

        temp.next = self.head
        self.head.prev = temp

    def insertAtBig(self, value):
        temp = Node(value)

        if self.head == None:
            self.head = temp
            temp.next = self.head
            temp.prev = self.head
            return

        t = self.head

        while t.next != self.head:
            t = t.next

        temp.next = self.head
        temp.prev = t

        self.head.prev = temp
        t.next = temp

        self.head = temp

    def insertAtMid(self, value, x):
        if self.head == None:
            return

        t = self.head

        while True:
            if t.data == x:
                break

            t = t.next

            if t == self.head:
                return

        temp = Node(value)

        temp.next = t.next
        temp.prev = t

        t.next.prev = temp
        t.next = temp

    def printDLL(self):
        if self.head == None:
            print("EMPTY")
            return

        t1 = self.head

        while True:
            print(t1.data, end=",")
            t1 = t1.next

            if t1 == self.head:
                break

        print()

    def deletionDLL(self, value):
        if self.head == None:
            print("EMPTY")
            return

        t = self.head

        # Only one node
        if t.next == self.head:
            if t.data == value:
                self.head = None
            return

        # Delete at beginning
        if t.data == value:
            last = self.head.prev

            self.head = self.head.next

            self.head.prev = last
            last.next = self.head

            return

        # Delete at mid / end
        t = self.head.next

        while t != self.head:

            if t.data == value:
                t.prev.next = t.next
                t.next.prev = t.prev
                return

            t = t.next


obj = DoublyLL()

obj.insertAtEnd(10)
obj.insertAtEnd(20)
obj.insertAtEnd(30)

obj.insertAtBig(6)

obj.insertAtMid(50, 20)

obj.deletionDLL(30)

obj.printDLL()

```

```python
class stack :
    def __init__(self):
        self.s =[]

    def length(self):
        return len(self.s)

    def push(self,value):
        self.s.insert(0,value)

    def peek(self):
        if len(self.s) == 0 :
            raise Exception("stack is empty")
        else:
            return self.s[0]

    def pop(self):
        if len(self.s) == 0 :
            raise Exception("stack is empty")
        else:
            return self.s.pop(0)

stk = stack()
stk.push(10)
stk.push(20)
stk.push(30)
stk.push(40)

print(stk.pop())
print(stk.pop())
print(stk.pop())
print(stk.pop())


    
```

Queue - FIFO

```python
class Queue:
    def __init__(self):
        self.items = []

    def isEmpty(self):
        return len(self.items) == 0

    def insert(self,value):
        self.items.append(value)

    def delete(self):
        if (self.isEmpty()):
            print('queue is empty')
        else:
            return self.items.pop(0)

Q = Queue()
Q.insert(10)
Q.insert(20)
Q.insert(30)

print(Q.delete())
print(Q.delete())


```

```python
# Double Ended Queue (Dequeue)

class Deque:
    def __init__(self):
        self.items = []

    def isEmpty(self):
        return len(self.items) == 0

    def insertAtEnd(self,value):
        self.items.append(value)

    def deleteAtFront(self):
        if (self.isEmpty()):
            print('queue is empty')
        else:
            return self.items.pop(0)

    def insertAtFront(self,value):
        self.items.insert(0,value)

    def deleteAtEnd(self):
        if (self.isEmpty()):
            print('queue is empty')
        else:
            return self.items.pop()
        

Q = Deque()
Q.insertAtEnd(10)
Q.insertAtFront(20)
Q.insertAtEnd(30)
Q.insertAtEnd(40)
Q.insertAtFront(50)

print(Q.deleteAtEnd())
print(Q.deleteAtEnd())
print(Q.deleteAtFront())
print(Q.deleteAtFront())
print(Q.deleteAtEnd())
Q.deleteAtFront()
Q.deleteAtFront()
```
