```python
# linear search
def  linear_search (arr,target):
    size = len(arr)
    for index in range(0,size):
        if (arr[index]==target):
            return index
    return -1

lists=[14,4,65,234,2,4,23,56,34,5]
target = 23

result= linear_search(lists,target)
print(result)
```

```python
# Binary search

def binary_search(arr,target):
    size= len(arr)

    start=0
    end=size -1

    while (start <= end):
        mid = (start + end)//2

        if(arr[mid]== target):
            return mid
        elif (arr[mid]> target):
            end =mid -1
        elif (arr[mid]< target):
            start =mid +1
    return -1

sorted_list = [10,20,44,23,54,38,34,23,22,53,35,83]
target=22 

result= binary_search(sorted_list,target)
print(result)
```
