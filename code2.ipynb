Functions : A function is a block of code that performed a specific task. Function help in organizing code, reusing code, and improving readability. 

```python
# why do requred function?
def oe (num):
    if num%2==0:
        print("even")
    else:
        print("odd")
oe(22)
```

```python
#  function with multiple parameters 
def add(a,b):
    return a+b
add(2,45)
```

```python
# default parameter
def greet(name):
    print(f"hello {name}")
greet("nav") 
```

```python
# variable length arguments 
# positional and keywords arguments
#positional: "Positional Argument woh argument hai jo function ko uske parameters ke same order (position) me pass kiya jata hai."
def print_numbers(*nav):
    for number in nav:
        print(number)
print_numbers(1,2,3,'my',4,6,3,43.4)
```

```python
# keyward arg: "Keyword Argument woh argument hai jisme function call karte waqt parameter ka naam likhkar value pass ki jati hai."
def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")
print_details(name="nav",age= "23")
```

Function Exmaple

```python
# Example 1: Temprature conversion
def convert_temprature(temp,unit):
    if unit== 'C':
        return temp * 9/5 +32 # celsius to fahreheit
    elif unit == 'F':
        return (temp-32) * 5/9 # F to C
    else:
        return None

print(convert_temprature(25,'C'))
print(convert_temprature(77,'F'))
```

```python
# passward Strength checker
def strong_passward(passward):
    if len(passward)<8:
        return False
    if not any(char.isdigit() for char in passward):
        return False
    if not any(char.islower() for char in passward):
        return False
    if not any(char.isupper() for char in passward):
        return False
    if not any(char in '!@#$%^&*' for char in passward):
        return False
    if passward == "abc123":
        print('easily crackable')
        return False
    return True
print(strong_passward(input("enter passward")))
```

```python
# example 3: calculate the total cost of item in shopping cart
def claculate_total_cost(cart):
    total_cost=0
    for items in cart:
        total_cost+=items['price']* items['quantity']
    return total_cost

cart=[
    {'name':'school bag', 'price':350,'quantity':+1},
    {'name':'shoes ', 'price':450,'quantity':+2},
    {'name':'books seat', 'price':150,'quantity':+6},
    {'name':'school dress', 'price':750,'quantity':+2}
]
total_cost=claculate_total_cost(cart)
print(total_cost)
```

```python
# Example 4: Check if a string is palindrome
def is_palindrome(s):
    s=s.lower().replace('','')
    return s==s[::-1]
print(is_palindrome(input('enter name')))
```

```python
# Example 5: calculate the factorials of a number using recursion
def fact(n):
    if n==0:
        return 1
    else:
        return n* fact(n-1)
print(fact(6))
```

```python
# example 6: A function to read a file and count the frequency of each word
def count_word_frequency(file_path):
    word_count={}
    with open(file_path,'r')as file:
        for line in file:
            words=line.split()
            for word in words:
                word=word.lower().strip('!@#$%^&*')
                word_count[word]=word_count.get(word,0)+1
    return word_count

filepath='for.py' 
word_frequency=count_word_frequency(filepath)
print(word_frequency)



```

lambda function : Lambda functions are small anonymous functions defined using a Lambda keyword. They can have any number of arguments but only one expression. They are commonly used for short operations or as arguments to higher order function .

```python
addition = lambda a,b,c : a+b+c
print(addition(3,5,2))
```

```python
even = lambda num: num%2==0
print(even(12))
```

```python
numbers=[1,2,3,4,5,6]
list(map(lambda x:x**2,numbers))
```

Map() function :- The map() function applies a given function to all the items in an input list (or any other iterable) and returns a map object (an iterator). This is particularly useful for transforming data in a list comprehensively.

```python
numbers=[1,2,3,4,5,6]
list(map(lambda a:a*a,numbers))
```

```python
N1=[1,2,3]
N2=[4,5,6]
add_num=list(map(lambda x,y:x+y,N1,N2))
print(add_num)
```

```python
str_num = ['1','2','3','4','5']
convert_int =list(map(int,str_num))
print(convert_int)
print(type(convert_int))
```

```python
str_num = ['apple','ball','cat','dog','elephant']
convert_upper =list(map(str.upper,str_num))
convert_upper1 =list(map(str.capitalize,str_num))
print(convert_upper)
print(convert_upper1)
```

filter() funtion : filter() Python ka built-in function hai jo kisi iterable (list, tuple, etc.) se sirf un elements ko return karta hai jo di gayi condition ko satisfy karte hain.

```python
lst = [1,2,3,4,5,6,7,8,9,3,4,6,4,5,54,3,4,4]
list(filter(even,lst))
```

```python
# filter with lambda function
largest_of_fives=list(map(lambda a:a>5,lst))
largest_of_five=list(filter(lambda a:a>5,lst))
print(largest_of_fives) 
print(largest_of_five) 
```

```python
even = list(filter(lambda x: x > 5 and x % 2 == 0, lst))
odd = list(filter(lambda x: x > 5 and x % 2 != 0, lst))

print("Even:", even)
print("Odd:", odd)
```
